```python
import pandas as pd
import os
import sys
from sklearn.model_selection import train_test_split

# 1. Load the massive 6.3M row dataset
# --- DYNAMIC PATH RESOLUTION ---
current_dir = os.getcwd()
if current_dir.endswith('notebooks'):
    project_root = os.path.dirname(current_dir)
else:
    project_root = current_dir

if project_root not in sys.path:
    sys.path.insert(0, project_root)

print(f"Project root set to: {project_root}")

raw_path = os.path.join(project_root, "data", "raw", "PS_20174392719_1491204439457_log.csv").replace('\\', '/')

print("Loading full PaySim dataset...")
df = pd.read_csv(raw_path)

# 2. Perform a stratified split to get exactly 100,000 rows
# 'isFraud' is the target column
print("Stratifying and sampling 100,000 rows...")
_, sample_df = train_test_split(
    df, 
    test_size=100000, 
    stratify=df['isFraud'], 
    random_state=42
)

# 3. Save the lightweight sample for your pipeline
output_path = os.path.join(project_root, "data", "raw", "paysim_100k_sample.csv").replace('\\', '/')
sample_df.to_csv(output_path, index=False)
print(f"✅ Stratified sample saved to {output_path}")
```

```python
import pandas as pd
import os
import sys

# --- DYNAMIC PATH RESOLUTION ---
current_dir = os.getcwd()
if current_dir.endswith('notebooks'):
    project_root = os.path.dirname(current_dir)
else:
    project_root = current_dir

if project_root not in sys.path:
    sys.path.insert(0, project_root)

print(f"Project root set to: {project_root}")
print("\n🔍 RUNNING PAYSIM SAMPLE DIAGNOSTIC...\n")

# Define precise paths
raw_path = os.path.join(project_root, "data", "raw", "PS_20174392719_1491204439457_log.csv").replace('\\', '/')
sample_path = os.path.join(project_root, "data", "raw", "paysim_100k_sample.csv").replace('\\', '/')

# 1. Fetch Original Ratio (Memory Efficient)
try:
    print("Fetching original ratio (loading only 'isFraud' column)...")
    df_orig = pd.read_csv(raw_path, usecols=['isFraud'])
    orig_fraud_pct = (df_orig['isFraud'].mean()) * 100
    print(f"Original Fraud Ratio: {orig_fraud_pct:.4f}%\n")
    
    # Free up memory immediately
    del df_orig 
except Exception as e:
    print(f"Could not load original: {e}\nSkipping original ratio check.\n")

# 2. Fetch Sample Data
print("Fetching sample dataset...")
df_sample = pd.read_csv(sample_path)

# 3. Print Diagnostics
print("\n" + "="*40)
print("--- DIAGNOSTIC RESULTS ---")
print("="*40)

print(f"Sample Shape: {df_sample.shape}")

sample_fraud_counts = df_sample['isFraud'].value_counts()
sample_fraud_pct = (df_sample['isFraud'].mean()) * 100

print(f"\nSample Target Distribution ('isFraud'):")
print(sample_fraud_counts.to_string())
print(f"Sample Fraud Ratio: {sample_fraud_pct:.4f}%")

print(f"\nMissing Values in Sample: {df_sample.isnull().sum().sum()}")
print("="*40)
```
