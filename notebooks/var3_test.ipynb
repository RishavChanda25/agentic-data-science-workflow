```python
import os
import sys
import time
from IPython.display import Markdown, display, Image
from dotenv import load_dotenv

# --- DYNAMIC PATH RESOLUTION ---
current_dir = os.getcwd()
if current_dir.endswith('notebooks'):
    project_root = os.path.dirname(current_dir)
else:
    project_root = current_dir

if project_root not in sys.path:
    sys.path.insert(0, project_root)

# Load environment variables
load_dotenv()

# Import the Variant 3 Dynamic Graph
from workflow_engine.orchestrators.dynamic_graph import build_dynamic_pipeline

# Import the Dataset Metadata Extractor
from workflow_engine.utils.data_profiler import extract_dataset_metadata

# NEW: Import the LLM Judge
from workflow_engine.utils.llm_judge import grade_report_with_llm
```

```python
# Setup Test Variables
raw_path = os.path.join(project_root, "data", "raw", "telco_customer_churn.csv").replace('\\', '/')
target_col = "Churn" # Update to match your CSV (num, isFraud, Churn)

user_request = f"I don't need a predictive model. I just want an executive-level understanding of this dataset. Produce exploratory analysis, key insights and visualisations that I can present to senior management without discussing machine learning algorithms."

# --- RUN THE PROFILER (Zero Tax) ---
print(" 🔍 [System] Running Pre-Flight Profiler...")
actual_metadata = extract_dataset_metadata(raw_path, target_col=target_col)
print(f" 📊 [System] Data Reality Detected:\n {actual_metadata}")

initial_state = {
    "messages": ["Starting Neurosymbolic Routing Test."],
    "user_request": user_request,
    "target_variable": target_col,
    "raw_dataset_path": raw_path,
    "current_dataset_path": raw_path,
    
    # --- VARIANT 3: DYNAMIC ROUTING INJECTION ---
    "active_preset": "",
    "proposed_route": [],
    "z3_verification_status": False,
    "dataset_metadata": actual_metadata,
    
    "artifacts": {},
    "current_step": "start",
    "error_flag": False,
    "error_message": "",
    "revision_count": 0,
    "user_preferences": {},
    "total_sleep_time": 0.0,
    "next_node": "",
    
    "total_input_tokens": 0,
    "total_output_tokens": 0,
    "api_call_timestamps": [],
    "supervisor_latency": 0.0,
    "supervisor_tokens": 0,
    "supervisor_calls": 0 
}

# Build the Graph
app = build_dynamic_pipeline()
print("✅ State Initialized and Graph Compiled.")
```

```python
from IPython.display import display, Markdown

print(" 🚀  Starting Variant 3 Neurosymbolic Execution...")
print(" ⏱️   Clock is running. Please wait...\n")

start_time = time.time()

try:
    final_state = app.invoke(initial_state, {"recursion_limit": 20})
except Exception as e:
    print(f" ❌  Graph execution crashed: {e}")
    final_state = {"error_flag": True, "error_message": str(e), "total_sleep_time": 0.0}

end_time = time.time()
gross_execution_time = end_time - start_time
total_sleep = final_state.get("total_sleep_time", 0.0)
pure_agent_execution_time = gross_execution_time - total_sleep

# --- PRINT THE TAX COMPARISON ---
if not final_state.get("error_flag"):
    
    # --- RETRIEVE STATE METRICS ---
    supervisor_latency = final_state.get("supervisor_latency", 0.0)
    supervisor_tokens = final_state.get("supervisor_tokens", 0)
    supervisor_calls = final_state.get("supervisor_calls", 0)
    
    timestamps = final_state.get("api_call_timestamps", [])
    in_tokens = final_state.get("total_input_tokens", 0)
    out_tokens = final_state.get("total_output_tokens", 0)
    total_tokens = in_tokens + out_tokens
    total_calls = len(timestamps)
    
    # --- CALCULATE WORKER VS SUPERVISOR SPLITS ---
    worker_execution_time = pure_agent_execution_time - supervisor_latency
    worker_tokens = total_tokens - supervisor_tokens
    worker_calls = total_calls - supervisor_calls
    
    print("\n" + "="*60)
    print(" 🧠  NEUROSYMBOLIC ROUTING RESULTS")
    print("="*60)
    print(f"Active Preset:   {final_state.get('active_preset')}")
    print(f"Executed Route:  {final_state.get('proposed_route')}")
    print(f"Z3 Verified:     {final_state.get('z3_verification_status')}")

    # --- 1. GLOBAL LATENCY METRICS ---
    print("\n--- ⏱️  GLOBAL LATENCY METRICS ---")
    print(f"Gross Wall-Clock Time:     {gross_execution_time:.2f} seconds")
    print(f"Total Artificial Sleep:    {total_sleep:.2f} seconds")
    print(f"Total Pure Compute Time:   {pure_agent_execution_time:.2f} seconds")
    
    # --- 2. THE ORCHESTRATION TAX (Supervisor) ---
    print("\n---  🏦  THE ORCHESTRATION TAX (Supervisor) ---")
    print(f"Supervisor Compute Time:   {supervisor_latency:.2f} seconds")
    print(f"Supervisor Token Tax:      {supervisor_tokens:,} tokens")
    print(f"Supervisor API Calls:      {supervisor_calls}")
    
    # --- 3. WORKER AGENT METRICS ---
    print("\n--- 👷  WORKER AGENT METRICS ---")
    print(f"Pure Worker Compute Time:  {worker_execution_time:.2f} seconds")
    print(f"Worker Token Cost:         {worker_tokens:,} tokens")
    print(f"Worker API Calls:          {worker_calls}")

    # --- 4. GLOBAL TOKEN & API TOTALS ---
    print("\n--- 🌍  GLOBAL TOKEN & API TOTALS ---")
    if total_calls > 1:
        total_api_time_mins = (timestamps[-1] - timestamps[0]) / 60.0
        rpm = total_calls / total_api_time_mins if total_api_time_mins > 0 else total_calls
        print(f"Effective RPM:             {rpm:.2f} requests/min")
    else:
        print(f"Total API Calls:           {total_calls}")

    print(f"Total Input Tokens:        {in_tokens:,}")
    print(f"Total Output Tokens:       {out_tokens:,}")
    print(f"Total Token Cost:          {total_tokens:,}")
    print("="*60)
    
    # --- 5. DISPLAY FINAL REPORT ---
    print("\n\n" + "="*60)
    print(" 📄  FINAL PIPELINE REPORT")
    print("="*60)
    
    # Dynamically locate and read the generated markdown report
    report_path = os.path.join(project_root, "reports", "final_reports", "final_report.md").replace('\\', '/')
    if os.path.exists(report_path):
        with open(report_path, "r", encoding="utf-8") as file:
            report_content = file.read()
        display(Markdown(report_content))
    else:
        print(f"⚠️ Report not found at expected path: {report_path}")

else:
    print(f"\n ⚠️  Pipeline failed: {final_state.get('error_message')}")
```

```python
# --- THE LLM JUDGE EVALUATION ---
print("\n" + "="*60)
print(" ⚖️  RUNNING INTERPRETABILITY JUDGE")
print("="*60)

if not final_state.get("error_flag"):
    report_path = os.path.join(project_root, "reports", "final_reports", "final_report.md").replace('\\', '/')
    
    if os.path.exists(report_path):
        print(f"📖 Reading final report from: {report_path}")
        with open(report_path, "r", encoding="utf-8") as file:
            report_content = file.read()
            
        print("🧠 Invoking OpenAI GPT-4o Judge...")
        
        # Call the Judge Utility
        judge_scores = grade_report_with_llm(
            report_text=report_content, 
            preset=final_state.get("active_preset", "UNKNOWN"), 
            user_request=initial_state["user_request"]
        )
        
        print("\n" + "="*60)
        print(" 🎓 LLM JUDGE REPORT CARD")
        print("="*60)
        print(f"📘 Technical Depth Score:  {judge_scores.get('Technical Score')}/10")
        print(f"🎯 Actionability Score:    {judge_scores.get('Actionability Score')}/10")
        print(f"🤝 Audience Alignment:     {judge_scores.get('Alignment Score')}/10")
        print(f"\n🧠 Judge Reasoning:\n{judge_scores.get('Judge Reasoning')}")
        print("============================================================\n")
    else:
        print(f"⚠️ Report not found at expected path: {report_path}. Cannot run Judge.")
else:
    print("⚠️ Pipeline failed or did not generate a final report. Cannot run Judge.")
```
